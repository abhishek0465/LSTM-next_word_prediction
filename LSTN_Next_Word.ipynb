#### Convert Data to Sequence

```python
sequences = tokenizer.texts_to_sequences(data)
```

```python
sequences[0]
```

### Convert sequence to supervised modelling problem

```python
input_sequence = []
for seq in sequences:
    if len(seq)<2:
        continue
    for i in range(1,len(seq)):
        ngram = seq[:i+1]
        input_sequence.append(ngram)
```

```python
max_len = max([len(i) for i in input_sequence])
```

```python
max_len
```

### Padding to craeet equal length sequences

```python
input = pad_sequences(input_sequence,padding='pre',maxlen=max_len)
```

```python
input[:3]
```

```python
x=input[:,:-1]
y=input[:,-1]
```

```python
x.shape
```

```python
y.shape
```

### Modelling Next Word Prediction Model

```python
vocab_size = len(tokenizer.word_index)+1
```

```python
vocab_size
```

```python
y = np.array(y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
max_len = x.shape[1] ### save number of timestamps on each sequence
```

```python
x.shape[1]
```

#### Model Creation

```python
model = Sequential([
    Embedding(input_dim=vocab_size,output_dim=128,input_shape=(max_len,)),
    LSTM(256),
    Dense(vocab_size,activation='softmax')
])
```

```python
model.summary()
```

```python
model2 = Sequential([
    Input(shape=(max_len,)),
    Embedding(input_dim=vocab_size,output_dim=128),
    LSTM(256),
    Dense(vocab_size,activation='softmax')
])
```

```python
model2.summary()
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
# This step is needed if Input layer is not used or input_shape npt passed into the sequential model
model.build(input_shape=(None,max_len))
```
